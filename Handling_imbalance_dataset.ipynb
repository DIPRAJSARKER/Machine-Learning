## How to Handle Imbalanced Data

### 1. Up Sampling

**Up Sampling** means increasing the number of observations in the **minority class**.

For example:

Before Up Sampling:

| Class | Samples |
|---|---:|
| Majority Class | 900 |
| Minority Class | 100 |

After Up Sampling:

| Class | Samples |
|---|---:|
| Majority Class | 900 |
| Minority Class | 900 |

Common techniques:

- Random Oversampling
- **SMOTE (Synthetic Minority Over-sampling Technique)**

### 2. Down Sampling

**Down Sampling** means reducing the number of observations in the **majority class**.

For example:

Before Down Sampling:

| Class | Samples |
|---|---:|
| Majority Class | 900 |
| Minority Class | 100 |

After Down Sampling:

| Class | Samples |
|---|---:|
| Majority Class | 100 |
| Minority Class | 100 |

Common technique:

- Random Undersampling

### Difference Between Up Sampling and Down Sampling

| Up Sampling | Down Sampling |
|---|---|
| Increases minority class | Decreases majority class |
| Does not remove majority samples | Removes some majority samples |
| Can increase dataset size | Can decrease dataset size |
| May cause overfitting | May cause information loss |

## Key Takeaway

> **Up Sampling → Increase the minority class**

> **Down Sampling → Decrease the majority class**

```python
import numpy as np
import pandas as pd 

# ser the random seed for reproducibility
np.random.seed(123)
# create dataframe with two clases

n_sample=1000
class_0_ratio=0.9
class_1_ratio=0.1
n_class_0=int(class_0_ratio*n_sample)
n_class_1=n_sample-n_class_0

print(n_class_0)
print(n_class_1)
```

```python
## create my dataframe with inblance dataset
np.random.seed(123)

class_0=pd.DataFrame({
    'feature_1':  np.random.normal(loc=0,scale=1,size=n_class_0),
    'feature_2':  np.random.normal(loc=0,scale=1,size=n_class_0),
    'target': [0]*n_class_0
})

class_1=pd.DataFrame({
    'feature_1':  np.random.normal(loc=0,scale=1,size=n_class_1),
    'feature_2':  np.random.normal(loc=0,scale=1,size=n_class_1),
    'target': [1]*n_class_1
})
```

```python
df=pd.concat([class_0,class_1]).reset_index(drop=True)
```

```python
df.tail()
```

```python
df['target'].value_counts()
```

```python
##Upsampling
df_majority=df[df['target']==0]
df_minority=df[df['target']==1]
```

```python
from sklearn.utils import resample
```

```python
df_minority_upsampled=resample(
    df_minority,replace=True,n_samples=len(df_majority),random_state=42
)
```

```python
df_upsampled=pd.concat([df_majority,df_minority_upsampled]).reset_index(drop=True)
```

```python
df_upsampled['target'].value_counts()
```

```python
#dowmsapling

df_majority_downsampled=resample(
    df_majority,replace=False,n_samples=len(df_minority),random_state=42 
)
```

```python
df_downsampled=pd.concat([df_majority_downsampled,df_minority]).reset_index(drop=True)
```

```python
df_downsampled['target'].value_counts()
```
